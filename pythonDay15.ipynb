methods of list - append, pop, index, insert, sort, remove, reverse, copy, count

```python
#append is used to add element at the last
l.append(10)
print(l)
```

```python
#pop is used to delete element from the list
l.pop()
print(l)
```

```python
#insert is used to add element in between the list
l.insert(2,9)
print(l)
```

```python
#index is used to check the index position
print(l)
l.index(7)
```

```python
#pop(index) is used to delete the particular element
l.pop(8)
print(l)
```
